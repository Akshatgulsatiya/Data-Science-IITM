```python
# Seaborn all datasets
sns.get_dataset_names()
```

```python
#Internet request
df_tatanic = sns.load_dataset('titanic')
df_tatanic.sample()
```

```python
# sns.countplot
# sns.barplot
# sns.boxplot
# sns.scatterplot
# sns.catplot
# sns.pairplot
# sns.heatmap
# sns.regplot
# sns.lmplot
# sns.lineplot
# sns.kdeplot
# sns.violinplot

# Major plots in seaborn.
```

```python
#sns.countplot()

import matplotlib.pyplot as plt
sns.countplot(data = df_tatanic,x = 'sex', hue = 'alive', palette=sns.color_palette('magma',2)) # hue : further divide :  based on categorical(object
plt.title('Titanic Survival Analysis basis on Gender')
plt.show()
```

```python
# sns.barlpot

sns.barplot(data = df_tatanic, x = 'class',y='fare',hue = 'alive',estimator='mean') # min,max,sum
plt.show()
```

```python
# sns.barlpot

sns.barplot(data = df_tatanic, x = 'class',y='fare',hue = 'alive',errorbar=('ci',0)) # min,max,sum
plt.grid(color = 'r', alpha = 0.1)
plt.show()
# Confidence Interval(CI) - Fluctuation of data 
```

```python
# sns.boxplot : To detect Outliers and to check data distribution 
# To detect outlier: 
# Range : min to max(19-22): 35,40,50(Outliers)
# Q2: Quantile : 50%
# Q1: 25%
# Q3: 75%

# # IQR = Q3 - Q1
# UW(Upper cap) = Q3 + 1.5*IQR
# LW(Lower cap) = Q1 - 1.5* IQR
```

```python
# sns.boxplot
sns.boxplot(data = df_tatanic,x ='class',y = 'fare',hue='sex')
plt.show()
```

```python
# sns.boxplot

plt.figure(figsize=(10,5))
sns.boxplot(data = df_tatanic,y ='class',x = 'fare',hue='sex',orient='h')
plt.show()
```

```python
# sns.violinplot

plt.figure(figsize=(10,5))
sns.violinplot(data = df_tatanic,x ='class',y = 'fare',hue='sex')
plt.show()
```

```python
df_pen = sns.load_dataset('penguins')
df_pen.sample()
```

```python
df_tips = sns.load_dataset('tips')
df_tips.sample()
```

```python
# sns.catplot 
# it allows multiple plot at once

sns.catplot(data= df_tips, x='time',y='total_bill', kind='violin',hue='sex',row = 'smoker',col = 'day')
plt.show()
```

```python
# sns.lmplot

sns.lmplot(data = df_tips,x='total_bill',y='tip',hue='sex',row='smoker',col='day')
plt.show()
```

```python
# sns.relplot 

sns.relplot(data = df_tips, x= 'total_bill', y = 'tip',kind = 'scatter', hue = 'sex', row = 'smoker', col='day')
plt.show()
```

```python
# sns.pairplot (numarical vs numarical)

df_iris = sns.load_dataset('iris')
df_iris.sample(5)
```

```python
df_iris.iloc[:,:-1].corr()
```

```python
# # >0: +ve : + ve Corr
# =0: No corr
# <0: -ve: - ve Corr
```

```python
sns.pairplot(data = df_iris, hue = 'species')
plt.show()
```

```python
# sns.kdeplot
# sns.jointplot|
```

```python
# sns.kdeplot : kernal density estimation

sns.kdeplot(data = df_iris, x = 'sepal_length', y = 'petal_width', hue = 'species')
sns.scatterplot(data = df_iris, x = 'sepal_length', y = 'petal_width', hue = 'species')
plt.show()
```

```python
# sns.kdeplot : Kernel density estimation

sns.kdeplot(data = df_iris, x = 'sepal_length', y = 'sepal_width', hue = 'species')
sns.scatterplot(data = df_iris, x = 'sepal_length', y = 'sepal_width', hue = 'species')
plt.show()
```

```python
# sns.kdeplot : Kernel density estimation

sns.jointplot(data = df_iris, x = 'sepal_length', y = 'sepal_width',kind = 'hex') # kind : { "scatter" | "kde" | "hist" | "hex" | "reg" | "resid" }
plt.show()
```

```python
# sns.kdeplot : Kernel density estimation

sns.jointplot(data = df_iris, x = 'sepal_length', y = 'sepal_width',kind = 'reg') # kind : { "scatter" | "kde" | "hist" | "hex" | "reg" | "resid" }
plt.show()
```
