```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#Machine learning
from sklearn import*
print('done')
```

```python
data = {'Study_hrs':range(1,11),
       'Marks Obtained':[25,33,46,50,55,67,78,85,95,100]}

df = pd.DataFrame(data)
df
```

```python
# ML model build?: x, y: +ve, -ve
# Scatter Plot
plt.title('Study vs Marks')
plt.scatter(df['Study_hrs'],df['Marks Obtained'],marker = 'x', color = 'r')
plt.xlabel('Study Hrs')
plt.ylabel('Marks')
plt.show()
```

```python
# # Y = mx + c
# # M : Slope
# # C : Intercept

# Y = Marks
# x = Study_hrs
```

```python
#x(1), y(1): simple linear Regression
#x(muliple): y(1): Multiple Linear Reegression
```

```python
# import module
from sklearn.linear_model import LinearRegression

# object model creation
model = LinearRegression()

# x,y
x=df[['Study_hrs']]
y=df['Marks Obtained']

# training model
model.fit(x,y)
```

```python
x = [[5.5]]

m = model.coef_
c = model.intercept_

print(m,c)
```

```python
# Y =  m*x + c
std_hrs = [[5.5]]

marks = m*std_hrs + c

print('Pridicted Marks:',marks[0][0])
```

```python
study_hrs = input('Enter study hours')
std_hrs = [[float(study_hrs)]]

model.predict(std_hrs)
ans = model.predict(std_hrs)
marks = round(ans[0],2) 
if marks > 100:
    marks =100
elif marks <0:
    marks = 0 
else:
    marks = marks   
print(f'Your predicted marks is : {marks}')
```
